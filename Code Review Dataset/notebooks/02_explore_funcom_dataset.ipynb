## EDA (Exploratory Data Analysis) of FunCom Dataset

1. Imports

```python
import os
from pathlib import Path
from tkinter import Tk, filedialog

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from collections import Counter
```

2. Select FunCom folder

```python
root = Tk()
root.withdraw()

funcom_dir = filedialog.askdirectory(
    title="Select funcom folder"
)

print("Selected:", funcom_dir)
```

3. Build folder tree

```python
def print_tree(path, prefix=""):
    entries = sorted(os.listdir(path))

    for i, entry in enumerate(entries):

        full_path = os.path.join(path, entry)

        connector = "└── " if i == len(entries)-1 else "├── "

        print(prefix + connector + entry)

        if os.path.isdir(full_path):

            extension = (
                "    "
                if i == len(entries)-1
                else "│   "
            )

            print_tree(full_path, prefix + extension)

print(Path(funcom_dir).name)
print_tree(funcom_dir)
```

4. Search all files

```python
def print_tree(path, prefix=""):
    entries = sorted(os.listdir(path))

    for i, entry in enumerate(entries):

        full_path = os.path.join(path, entry)

        connector = "└── " if i == len(entries)-1 else "├── "

        print(prefix + connector + entry)

        if os.path.isdir(full_path):

            extension = (
                "    "
                if i == len(entries)-1
                else "│   "
            )

            print_tree(full_path, prefix + extension)

print(Path(funcom_dir).name)
print_tree(funcom_dir)
```

5. Search necessary files

```python
all_files = []

for root_dir, dirs, files in os.walk(funcom_dir):
    for file in files:
        all_files.append(os.path.join(root_dir, file))

print("Files found:", len(all_files))

for f in all_files:
    print(f)
```

6. Find Standard Dataset

```python
standard_output = os.path.join(
    funcom_dir,
    "data",
    "standard",
    "output"
)

print(standard_output)
```

7. Look up files

```python
for file in sorted(os.listdir(standard_output)):
    print(file)
```

8. First rows of files

```python
files_to_check = [
    "dataset.coms",
    "dataset.dats",
    "dataset.srcml"
]

for fname in files_to_check:

    path = os.path.join(
        standard_output,
        fname
    )

    print("\n")
    print("="*80)
    print(fname)

    with open(
        path,
        encoding="utf8",
        errors="ignore"
    ) as f:

        for _ in range(3):
            print(f.readline())
```

9. Download Train

```python
with open(
    os.path.join(
        standard_output,
        "coms.train"
    ),
    encoding="utf8"
) as f:

    train_comments = f.read().splitlines()

with open(
    os.path.join(
        standard_output,
        "dats.train"
    ),
    encoding="utf8"
) as f:

    train_code = f.read().splitlines()

with open(
    os.path.join(
        standard_output,
        "smls.train"
    ),
    encoding="utf8"
) as f:

    train_ast = f.read().splitlines()

print(len(train_comments))
print(len(train_code))
print(len(train_ast))
```

10. Train DataFrame

```python
train_df = pd.DataFrame({
    "code": train_code,
    "comment": train_comments,
    "ast": train_ast
})

train_df.head()
```

11. Validation

```python
def load_split(split):

    with open(
        os.path.join(
            standard_output,
            f"coms.{split}"
        ),
        encoding="utf8"
    ) as f:

        comments = f.read().splitlines()

    with open(
        os.path.join(
            standard_output,
            f"dats.{split}"
        ),
        encoding="utf8"
    ) as f:

        code = f.read().splitlines()

    with open(
        os.path.join(
            standard_output,
            f"smls.{split}"
        ),
        encoding="utf8"
    ) as f:

        ast = f.read().splitlines()

    return pd.DataFrame({
        "code": code,
        "comment": comments,
        "ast": ast
    })

val_df = load_split("val")
test_df = load_split("test")

print(train_df.shape)
print(val_df.shape)
print(test_df.shape)
```

15. Length of code

```python
train_df["code_tokens"] = (
    train_df["code"]
    .str.split()
    .apply(len)
)

train_df["comment_tokens"] = (
    train_df["comment"]
    .str.split()
    .apply(len)
)

train_df["ast_tokens"] = (
    train_df["ast"]
    .str.split()
    .apply(len)
)
```
